```python
import numpy as np
import pandas as pd 

import w3_unittest
import utils
```

```python
paramters = utils.initialize_parameters(2)
```

```python
def forward_propagation(X, parameters):
    W = parameters["W"]
    b = parameters["b"]
    Z = np.dot(W, X) + b
    Y_hat = Z

    return Y_hat
```

```python
w3_unittest.test_forward_propagation(forward_propagation)
```

```python
def compute_cost(Y_hat, Y):
    m = Y.shape[1]

    cost = np.sum((Y_hat - Y)**2)/ (2*m)
    return cost
```

```python
def nn_model(X, Y, num_iterations=1000, print_cost= False):
    n_x = X.shape[0] # considering the features 

    parameters = utils.initialize_parameters(n_x)

    for i in range(0, num_iterations):
        Y_hat = forward_propagation(X, parameters)
        cost = compute_cost(Y_hat, Y)

        parameters = utils.train_nn(parameters, Y_hat, X, Y, learning_rate = 0.001)
        
        if print_cost:
            if i%100 ==0:
                print("Cost after iteration is %i: %f" %(i, cost))
    return parameters    
```

```python
w3_unittest.test_nn_model(nn_model)
```
